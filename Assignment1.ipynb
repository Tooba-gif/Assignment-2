# ***Mad Libs Game in Python***


```python
#user information
name = input("Enter the name;")
programming_language = input("Enter the programmer language (e.g python , typescript);")
mentor_name = input ("Enter the mentor name;")
favourite_food = input("Enter the favourite food ")
feeling = input("Enter the feeling;")

#story created
print(" Story behind the Mad Libs Game!")
print(f"I am {name}, a developer learning {programming_language} from my mentor, {mentor_name}.")
print(f"Yesterday, I coded a project that solved a complex problem. I was feeling very {feeling} and decided to celerate with {favourite_food}.")
```
